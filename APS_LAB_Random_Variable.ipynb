```python
import numpy as np
import matplotlib.pyplot as plt

# Theoretical probabilities
theoretical_pmf = [0.7, 0.3]

# Empirical probabilities obtained from the sample
empirical_pmf = [0.65, 0.35]

# Outcomes
x = [0, 1]

bar_width = 0.35
index = np.arange(len(x))

# Plot theoretical PMF
plt.bar(index, theoretical_pmf, bar_width, label='Theoretical')

# Plot empirical PMF
plt.bar(index + bar_width, empirical_pmf, bar_width, label='Empirical')

# Labels and title
plt.xlabel('Outcome')
plt.ylabel('Probability')
plt.title('Theoretical vs Empirical PMF')

# X-axis labels
plt.xticks(index + bar_width / 2, ['No Click', 'Click'])

# Show legend
plt.legend()

# Display graph
plt.show()
```

```python
import numpy as np
import pandas as pd

p = 0.30
sample_sizes = [10, 100, 1000, 10000]

theoretical_p0 = 1 - p
theoretical_p1 = p
theoretical_mean = p
theoretical_variance = p * (1 - p)

results = []

for size in sample_sizes:
    sample = np.random.binomial(n=1, p=p, size=size)

    empirical_p0 = np.sum(sample == 0) / size
    empirical_p1 = np.sum(sample == 1) / size
    empirical_mean = np.mean(sample)
    empirical_variance = np.var(sample)

    results.append({
        'Sample size': size,
        'Empirical P(X=0)': f'{empirical_p0:.4f}',
        'Empirical P(X=1)': f'{empirical_p1:.4f}',
        'Empirical mean': f'{empirical_mean:.4f}',
        'Empirical variance': f'{empirical_variance:.4f}'
    })

df_results = pd.DataFrame(results)
display(df_results)

print(f"\nTheoretical values:\nP(X=0) = {theoretical_p0:.2f}\nP(X=1) = {theoretical_p1:.2f}\nMean = {theoretical_mean:.2f}\nVariance = {theoretical_variance:.2f}")
```

Observation :
As the sample size increases from 10 to 10,000, the empirical probabilities, mean, and variance generally become closer to their theoretical values of 0.70, 0.30, 0.30, and 0.21 respectively. This demonstrates that larger samples provide a better approximation to the theoretical distribution, illustrating the Law of Large Numbers.

Problem statement: A fraud-detection model examines 50 transactions every hour. Assume that each transaction has a 4% probability of being flagged as potentially fraudulent.

Let: X = "number of transactions flagged in one hour"

Initially assume that individual flagging decisions are independent.

The fraud investigation team can examine at most four flagged transactions per hour.

Find:

Simulate the operation of the fraud-detection system for 20,000 independent one-hour periods. During every simulated hour:
exactly 50 transactions are examined;
each transaction has a flagging probability of 0.04;
X records the total number of flagged transactions.
Hint: use np.random.binomial function

```python
import numpy as np

# Define the parameters for the binomial distribution
n_transactions = 50  # Number of trials (transactions per hour)
p_flagged = 0.04     # Probability of a transaction being flagged
n_simulations = 20000 # Number of independent one-hour periods to simulate

# Simulate the number of flagged transactions for each one-hour period
# np.random.binomial(n, p, size) generates 'size' number of samples
# where each sample is the number of successes (flagged transactions)
# in 'n' trials (transactions) with 'p' probability of success.
flagged_transactions = np.random.binomial(n=n_transactions, p=p_flagged, size=n_simulations)

print(f"Simulated {n_simulations} one-hour periods.")
print(f"First 10 simulated values for flagged transactions: {flagged_transactions[:10]}")
print(f"Minimum flagged transactions in a period: {np.min(flagged_transactions)}")
print(f"Maximum flagged transactions in a period: {np.max(flagged_transactions)}")
```
