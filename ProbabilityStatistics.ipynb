```python
import random
# First question
num_tosses=100
result=random.choices(("Head","Tail"),k=num_tosses)
num_tails=result.count("Tail")
num_heads=result.count("Head")
print("Number of tails : ",num_tails)
print("Number of heads : ",num_heads)
print("Probability of  heads :" ,num_heads/num_tosses)
print("Probability of tails :" ,num_tails/num_tosses)
```

```python
trails=[10,100,1000,10000]
for trail in trails:
    result=random.choices(("Head","Tail"),k=trail)
    num_tails=result.count("Tail")
    num_heads=result.count("Head")
    print("Number of trails : ",trail)
    print("Number of tails : ",num_tails)
    print("Number of heads : ",num_heads)
    print("Probability of  heads :" ,num_heads/trail)
    print("Probability of tails :" ,num_tails/trail)
    print("*****************************")
```

```python
# A fair die is rolled 6,000 times.
#Calculate the experimental probability of each outcome: S={1,2,3,4,5,6}
num_trails=6000
result=random.choices(range(1,7),k=num_trails)
num_of_1s=result.count(1)
num_of_2s=result.count(2)
num_of_3s=result.count(3)
num_of_4s=result.count(4)
num_of_5s=result.count(5)
num_of_6s=result.count(6)
prob_of_1s=num_of_1s/num_trails
prob_of_2s=num_of_2s/num_trails
prob_of_3s=num_of_3s/num_trails
prob_of_4s=num_of_4s/num_trails
prob_of_5s=num_of_5s/num_trails
prob_of_6s=num_of_6s/num_trails
print("Probability of 1 : ",prob_of_1s)
print("Probability of 2 : ",prob_of_2s)
print("Probability of 3 : ",prob_of_3s)
print("Probability of 4 : ",prob_of_4s)
print("Probability of 5 : ",prob_of_5s)
print("Probability of 6 : ",prob_of_6s)
```

```python
# Addition rule for disjoint events
# A die is rolled 10,000 times.
# Define: A=getting 1, B= getting 6, check the events are joint or disjoint
# Verify: P(A∪B)=P(A)+P(B)
num_trails=10000
result=random.choices(range(1,7),k=num_trails)
num_of_1s=result.count(1)
num_of_6s=result.count(6)
prob_of_1s=num_of_1s/num_trails
prob_of_6s=num_of_6s/num_trails
probability_of_union=prob_of_1s+prob_of_6s
print("Probability of 1 : ",prob_of_1s)
print("Probability of 6 : ",prob_of_6s)
print("Probability of union : ",probability_of_union)

```

```python
# Addition rule for overlapping events
# For one die roll, define: A=getting an even number A={2,4,6} and B=getting a
# number greater than 3 B={4,5,6}
# Verify: P(A∪B)=P(A)+P(B)−P(A∩B)
num_trails=1
result=random.choices(range(1,7),k=num_trails)
print(result)
num_A = result.count(2)+result.count(4)+result.count(6)
num_B = result.count(4)+result.count(5)+result.count(6)
num_A_and_B = result.count(6)
prob_of_A = num_A/num_trails
prob_of_B = num_B/num_trails
prob_of_A_and_B = num_A_and_B/num_trails
probability_of_union = prob_of_A + prob_of_B - prob_of_A_and_B
print("Probability of A : ",prob_of_A)
print("Probability of B : ",prob_of_B)
print("Probability of A and B : ",prob_of_A_and_B)
print("Probability of union : ",probability_of_union)
```

```python
# Multiplication rule for independent events
# A coin is tossed and a die is rolled 10,000 times.
# Define:
#  obtaining Head
# obtaining an even number
# Verify: P(A∩B)=P(A)P(B)
two,four,six=0,0,0
head=0
tail=0
number=10000
for i in range(number):
  roll=random.randint(1,6)
  toss=random.choice(['HEAD','TAIL'])
  if roll==2:
    two+=1
  elif roll==4:
    four+=1
  elif roll==6:
    six+=1
  elif toss=="HEAD":
    head+=1
  elif toss=="tail":
    tail+=1

event_A=two+four+six
event_B=head
prob_of_event_A=event_A/number
prob_of_event_B=event_B/number
print(f"the  intersection of the event A {prob_of_event_A} and Event B {prob_of_event_B} is  {prob_of_event_A*prob_of_event_B}  ")
```
